# YOLO Training Pipeline

This notebook handles the training of the YOLO model for Airport Tools detection. 
It includes integration with MLflow for experiment tracking and configured data augmentations.

```python
import os
from ultralytics import YOLO, settings
import mlflow

# Enable MLflow integration in Ultralytics
settings.update({'mlflow': True})

# Check Ultralytics settings
print(settings)
```

## Configuration
Define paths, training parameters, and augmentations here.

```python
# Paths
DATA_YAML = "/home/jupyter/datasphere/filestore/datasets/Airport-Tools/data.yaml"
PROJECT_DIR = "/home/jupyter/datasphere/filestore/training/runs/"
MODEL_PATH = "yolo11x.pt" # Or path to your pretrained model

# MLflow Configuration
MLFLOW_TRACKING_URI = "http://127.0.0.1:5000" # REPLACE with your MLflow server URI
EXPERIMENT_NAME = "airport_tools_yolo_v1"

# Training Parameters
TRAIN_PARAMS = {
    "data": DATA_YAML,
    "project": PROJECT_DIR,
    "name": EXPERIMENT_NAME,
    "epochs": 50,
    "batch": 16, # Adjust based on GPU memory
    "imgsz": 640,
    "device": 0,
    "optimizer": "auto",
    "patience": 10,
    "save": True,
    "save_period": 5,
    "exist_ok": True,
    "verbose": True,
    "seed": 42,
    
    # Augmentations (Hyperparameters)
    "degrees": 0.0,      # Image rotation (+/- deg)
    "translate": 0.1,    # Image translation (+/- fraction)
    "scale": 0.5,        # Image scale (+/- gain)
    "shear": 0.0,        # Image shear (+/- deg)
    "perspective": 0.0,  # Image perspective (+/- fraction), range 0-0.001
    "flipud": 0.0,       # Image flip up-down (probability)
    "fliplr": 0.5,       # Image flip left-right (probability)
    "mosaic": 1.0,       # Image mosaic (probability)
    "mixup": 0.0,        # Image mixup (probability)
    "copy_paste": 0.0,   # Segment copy-paste (probability)
    "erasing": 0.4,      # Random erasing during classification training
    "crop_fraction": 1.0 # Image crop fraction
}
```

## Training
Initialize the model and start training.

```python
# Set MLflow Tracking URI and Experiment
mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)
mlflow.set_experiment(EXPERIMENT_NAME)

# Initialize Model
model = YOLO(MODEL_PATH)

# Start Training
results = model.train(**TRAIN_PARAMS)
```

## Validation
The model is automatically validated at the end of training. 
However, you can run additional validation here if needed.

```python
metrics = model.val()
```
